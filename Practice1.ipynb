```python
target=40
score=0
overs=0
runs_per_over=[9,6,12,4,11,7]
for runs in runs_per_over:
  score=score+runs
  overs=overs+1
  if score>=target:
    break
  print("overs", overs,"score",score)
if score>=target:
  print("Target reached in" , overs,"overs")
else:
  print("Target not reached. Final score", score)
```

```python
raw_plates = ["  tn09ab1234 ", "ka05mn4821", " TN22cd0057", "mh12ef9910  ", "tn07gh3345"]
clean_plates=[]
rto_codes=[]
reverse_order=[]
for number in raw_plates:
  clean_plates.append(number.strip().upper())
print("Total Vehicles : ", len(clean_plates))
print("Clean Plates : " ,clean_plates)
for number in clean_plates:
   rto_codes.append(number[2:4])
print("RTO Codes Sorted : ", sorted(rto_codes))
reg_no=(clean_plates[-1])
print("Last Vehicle number : ",reg_no[6::])

for number in clean_plates:
  reverse_order.append(number[::-1])
print(reverse_order)

```
